```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Superstore.csv', encoding='latin1')
df.head()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
df2 = df.drop(columns=['Order ID','Customer ID','Customer Name', 'Product ID', 'Postal Code','Product Name'], axis=1)
```

```python
df2['Sales.2'] = np.floor(df['Sales'] * 100) / 100
df2
```

```python
df2['Profit.2'] = np.floor(df['Profit'] * 100) /100
df2
```

```python
df2.describe()
```

```python
df3 = df2.drop(columns=['Sales','Profit'])
df3
```

```python
filtered_df = df3[df3['Sales.2'] < 500] # the max sales 499.99
filtered_df.max()
```

```python
plt.Figure(figsize=(16,8))

plt.subplot(1,2,1)
sns.histplot(x=df3['Sales.2'], bins=50, kde=True)
plt.title('Befor clean')

plt.subplot(1,2,2)
sns.histplot(x=filtered_df['Sales.2'], bins=50, kde=True)
plt.title('After clean')

plt.tight_layout()
plt.show()
```

```python
rows_lost = len(df3) - len(filtered_df)
percent_lost = (rows_lost / len(df3)) * 100
print(f'= {percent_lost:.2f}%')
```

```python
coutoff = df3['Sales.2'].quantile(0.95)
df_after_coutoff_95 = df3[df3['Sales.2']< coutoff]
df_after_coutoff_95.shape
```

```python
df_after_coutoff_95.describe()
```

```python
df4 = df_after_coutoff_95
```

```python
df4.describe()
```

```python
df_1 = df4[df4['Sales.2'] > 500]
df_2 = df4[df4['Sales.2'] <= 500]
```

```python
columns_drop = ['Row ID', 'Order Date', 'Ship Date','Country','City', 'State','Sub-Category', 'Quantity','Discount', 'New_Sales', 'New_Profit']
data_Visullization_1 = df_1.drop(columns_drop, axis=1)
# Sub-Category
data_Visullization_1
```

```python
data_Visullization_2 = df_2.drop(columns_drop, axis=1)
# Sub-Category
data_Visullization_2
```

```python
for col in data_Visullization_1.columns:
    plt.Figure()
    sns.barplot(x=data_Visullization_1[col], y=df_1['Sales.2'])
    plt.xlabel(col)
    plt.ylabel('Sales')
    plt.show()
```

```python
for col in data_Visullization_1.columns:
    plt.Figure()
    sns.boxplot(x=data_Visullization_1[col], y=df_1['Sales.2'])
    plt.xlabel(col)
    plt.ylabel('Sales')
    plt.show()
```

```python
for col in data_Visullization_2.columns:
    plt.Figure()
    sns.barplot(x=data_Visullization_2[col], y=df_2['Sales.2'])
    plt.xlabel(col)
    plt.ylabel('Sales')
    plt.show()
```

```python
for col in data_Visullization_2.columns:
    plt.Figure()
    sns.boxplot(x=data_Visullization_2[col], y=df_2['Sales.2'])
    plt.xlabel(col)
    plt.ylabel('Sales')
    plt.show()
```

```python
df4['Order Date'] = pd.to_datetime(df4['Order Date'])

numeric_cols = df4.select_dtypes(include='number').columns

monthly_sales = df4.groupby(df4['Order Date'].dt.to_period('M'))[numeric_cols].sum().reset_index()

monthly_sales['Order Date'] = monthly_sales['Order Date'].dt.to_timestamp()

sns.lineplot(x='Order Date', y='Sales.2', data=monthly_sales)
plt.xticks(rotation=45)
plt.title('monthly sales')
plt.show()
```

```python
df4['Ship Date'] = pd.to_datetime(df4['Ship Date'])

numeric_cols = df4.select_dtypes(include='number').columns

monthly_sales = df4.groupby(df4['Ship Date'].dt.to_period('M'))[numeric_cols].sum().reset_index()

monthly_sales['Ship Date'] = monthly_sales['Ship Date'].dt.to_timestamp()

sns.lineplot(x='Ship Date', y='Sales.2', data=monthly_sales)
plt.xticks(rotation=45)
plt.title('monthly sales')
plt.show()
```
